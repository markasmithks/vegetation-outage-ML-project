```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

n = 750

df = pd.DataFrame({
    "Year": rng.choice([2023, 2024], n),
    "Duration": rng.gamma(2.0, 90, n).round(1),
    "Customers_Out": rng.lognormal(3.0, 1.0, n).astype(int),
    "District": rng.choice(["North", "South"], n),
    "Substation": rng.choice(
        ["SUB001", "SUB002", "SUB003", "SUB004"], n
    ),
    "Feeder": rng.choice(
        ["FDR01", "FDR02", "FDR03", "FDR04"], n
    )
})

df["Customer_Minutes"] = (
    df["Duration"] * df["Customers_Out"]
).round()

df["Substation_Feeder"] = (
    df["Substation"] + "_" + df["Feeder"]
)

df.head()
```
